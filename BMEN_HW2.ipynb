```python
#Homework 2 - Question 2 part c and d
import numpy as np
import matplotlib.pyplot as plt
#constants
w = 0.5
af=0.25
vmax_rel=3 #l0/s
strong_l0=10 #cm
stretch_l0=20 #cm
f0_strong = 1800 #N
f0_stretch = 900 #N

#variable
l_strong = np.linspace(0.5 * strong_l0, 1.5 * strong_l0, 100)
l_stretch = np.linspace(0.5 * stretch_l0, 1.5 * stretch_l0, 100)
vmax_strong = strong_l0*vmax_rel
vmax_stretch = stretch_l0*vmax_rel
normal_strong=l_strong/strong_l0
normal_stretch=l_stretch/stretch_l0
v_strong = np.linspace(0, vmax_strong, 100)
v_stretch = np.linspace(0, vmax_stretch, 100)

# Normalized Velocities (0 to 1)
v_norm_strong = v_strong / vmax_strong
v_norm_stretch = v_stretch / vmax_stretch

#Force-Length
fl_strong = f0_strong* (1-((normal_strong-1)/w)**2)
fl_stretch = f0_stretch*(1-((normal_stretch-1)/w)**2)

#Force-Velocity
fv_strong = f0_strong*((1-v_norm_strong)/(1+(v_norm_strong/af)))
fv_stretch = f0_stretch*((1-v_norm_stretch)/(1+(v_norm_stretch/af)))

#Power-Velocity
p_strong = fv_strong*v_strong
p_stretch = fv_stretch*v_stretch

#Plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Force-Length Plot
axes[0].plot(l_strong, fl_strong, label="Strong muscle", linewidth=2)
axes[0].plot(l_stretch, fl_stretch, label="Stretch muscle", linewidth=2)
axes[0].set_title("Force–Length Curve")
axes[0].set_xlabel("Muscle length (cm)")
axes[0].set_ylabel("Force (N)")
axes[0].grid(True)
axes[0].legend()

# Force-Velocity Plot
axes[1].plot(v_strong, fv_strong, label="Strong muscle", linewidth=2)
axes[1].plot(v_stretch, fv_stretch, label="Stretch muscle", linewidth=2)
axes[1].set_title("Force–Velocity Curve")
axes[1].set_xlabel("Shortening velocity (cm/s)")
axes[1].set_ylabel("Force (N)")
axes[1].grid(True)
axes[1].legend()

# Power-Velocity Plot
axes[2].plot(v_strong, p_strong, label="Strong muscle", linewidth=2)
axes[2].plot(v_stretch, p_stretch, label="Stretch muscle", linewidth=2)
axes[2].set_title("Power–Velocity Curve")
axes[2].set_xlabel("Shortening velocity (cm/s)")
axes[2].set_ylabel("Power (N·cm/s)")
axes[2].grid(True)
axes[2].legend()

plt.tight_layout()
plt.show()
```
